## Inspect 'After Waive' Column

### Subtask:
Examine the unique values in the `After Waive` column to understand its content and identify the value(s) representing 'Pass'. This step is crucial to correctly interpret the 'Pass' criteria.


**Reasoning**:
To understand the content of the 'After Waive' column and identify values representing 'Pass', I will retrieve and display the unique values and their counts.



```python
after_waive_counts = df['After Waive'].value_counts()
display(after_waive_counts)
```

## Calculate 'Pass' Metrics per Team

### Subtask:
Group the DataFrame by the 'ASSIGN_TO' column. For each team, calculate the total number of assigned jobs, the number of jobs that are considered 'Pass' based on the `After Waive` column, and the percentage of 'Pass' jobs.


**Reasoning**:
To calculate the total number of jobs assigned to each team, I will group the DataFrame by the 'ASSIGN_TO' column and then count the occurrences within each group.



```python
total_jobs = df.groupby('ASSIGN_TO').size().rename('Total Jobs')
display(total_jobs)
```

**Reasoning**:
To calculate the number of 'Pass' jobs for each team, I will filter the DataFrame to include only rows where 'After Waive' is 'Pass', and then group by 'ASSIGN_TO' to count these jobs.



```python
pass_jobs = df[df['After Waive'] == 'Pass'].groupby('ASSIGN_TO').size().rename('Pass Jobs')
display(pass_jobs)
```

**Reasoning**:
To combine the calculated metrics into a single DataFrame, I will use `pd.DataFrame` to merge the `total_jobs` and `pass_jobs` Series, and then calculate the 'Pass Percentage' as a new column.



```python
team_performance = pd.DataFrame({
    'Total Jobs': total_jobs,
    'Pass Jobs': pass_jobs
}).fillna(0) # Fill NaN with 0 for teams with no pass jobs or no total jobs

team_performance['Pass Percentage'] = (team_performance['Pass Jobs'] / team_performance['Total Jobs'] * 100).fillna(0) # Handle division by zero if any
display(team_performance)
```

## Display Results Table

### Subtask:
Present these calculated metrics (total jobs, pass jobs, pass percentage) in a clear table format, showing the performance of each 'ASSIGN_TO' team.


**Reasoning**:
To present the calculated metrics, I will display the `team_performance` DataFrame, which already contains the 'Total Jobs', 'Pass Jobs', and 'Pass Percentage' for each team.



```python
display(team_performance)
```

```python
plt.figure(figsize=(12, 7)) # Increased figure width for better label spacing
ax = sns.barplot(x=team_performance['Pass Percentage'], y=team_performance.index, hue=team_performance.index, palette='viridis', legend=False)
plt.xlabel('เปอร์เซ็นต์งานที่ผ่าน') # Pass Percentage
plt.ylabel('ทีมที่ถูก Assign') # Assigned Team
plt.title('เปอร์เซ็นต์งานที่ผ่านของแต่ละทีม') # Pass Percentage of Each Team

# Add numerical labels to the end of each bar
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%', padding=5)

plt.xlim(0, 105) # Adjust x-axis limit to make space for labels
plt.tight_layout()
plt.show()
```

### 10. สร้างกราฟแท่งแบบ Stacked Bar Chart แสดงจำนวนงาน 'Pass' และ 'Not Waive' ในแต่ละทีม

```python
# Calculate 'Not Waive' jobs
team_performance['Not Waive Jobs'] = team_performance['Total Jobs'] - team_performance['Pass Jobs']

# Prepare data for stacked bar chart
# We need to unstack the 'Pass Jobs' and 'Not Waive Jobs' for plotting
plot_data = team_performance[['Pass Jobs', 'Not Waive Jobs']].reset_index()
plot_data = plot_data.melt(id_vars='ASSIGN_TO', var_name='Status', value_name='Count')

display(plot_data)
```

จากนั้นเราจะนำข้อมูลที่เตรียมไว้มาสร้างกราฟแท่งแบบ Stacked Bar Chart ค่ะ

```python
plt.figure(figsize=(12, 7))
ax = sns.barplot(x='ASSIGN_TO', y='Count', hue='Status', data=plot_data, palette={'Pass Jobs': 'lightgreen', 'Not Waive Jobs': 'salmon'})

# Set primary axis labels and title
ax.set_xlabel('ทีมที่ถูก Assign')
ax.set_ylabel('จำนวนงาน')
ax.set_title('จำนวนงาน Pass และ Not Waive ของแต่ละทีม พร้อมเปอร์เซ็นต์งานที่ผ่านและเส้นเป้าหมาย')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability

# Add numerical labels to each bar segment
for container in ax.containers:
    labels = [f'{int(v.get_height())}' if v.get_height() > 0 else '' for v in container]
    ax.bar_label(container, labels=labels, label_type='center', color='black', fmt='%s')

# Create a secondary y-axis for the percentage line plot
ax2 = ax.twinx()
ax2.plot(team_performance.index, team_performance['Pass Percentage'], color='blue', marker='o', linestyle='-', label='เปอร์เซ็นต์งานที่ผ่าน')
ax2.set_ylabel('เปอร์เซ็นต์งานที่ผ่าน (%)', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')
ax2.set_ylim(0, 100) # Set y-axis limit for percentage

# Add numerical labels to the line plot points
for i, percentage in enumerate(team_performance['Pass Percentage']):
    ax2.text(i, percentage + 2, f'{percentage:.2f}%', color='blue', ha='center')

# Add a target line at 90%
target_percentage = 90
ax2.axhline(y=target_percentage, color='red', linestyle='--', label=f'เป้าหมาย ({target_percentage}%)')
ax2.text(len(team_performance.index) - 0.5, target_percentage + 2, f'เป้าหมาย {target_percentage}%', color='red', va='center', ha='right')

# Combine legends from both axes
lines, labels = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='upper left')

plt.tight_layout()
plt.show()
```

กราฟแท่งแบบ Stacked Bar Chart นี้จะแสดงให้เห็นว่าในแต่ละทีมมีจำนวนงานที่ 'Pass' และ 'Not Waive' เป็นจำนวนเท่าใด ทำให้เห็นสัดส่วนและปริมาณของงานแต่ละประเภทได้อย่างชัดเจนค่ะ

```python
filtered_df = df[df['After Waive'].isin(['Not Waive', 'Pass'])]
print("Filtered DataFrame shape:")
print(filtered_df.shape)
```

```python
grouped_counts = filtered_df.groupby(['ASSIGN_TO', 'After Waive']).size().unstack(fill_value=0)
print("Grouped counts of 'Not Waive' and 'Pass' per ASSIGN_TO:")
print(grouped_counts.head())
```

```python
total_counts = grouped_counts.sum(axis=1)
percentages = grouped_counts.divide(total_counts, axis=0) * 100

# Rename columns for clarity as per instruction #7
percentages.columns = [col + '_Percentage' for col in percentages.columns]

print("Percentage distribution of 'Not Waive' and 'Pass' per ASSIGN_TO:")
print(percentages.head())
```

```python
percentages_sorted = percentages.sort_values(by='Pass_Percentage', ascending=False)

# Reorder columns to have 'Pass_Percentage' at the bottom and 'Not Waive_Percentage' on top
percentages_to_plot = percentages_sorted[['Pass_Percentage', 'Not Waive_Percentage']]

plt.figure(figsize=(15, 10))
ax = percentages_to_plot.plot(kind='bar', stacked=True, figsize=(15, 10), colormap='viridis')
plt.title('Percentage Distribution of "Not Waive" and "Pass" per ASSIGN_TO')
plt.xlabel('ASSIGN_TO')
plt.ylabel('Percentage')
plt.xticks(rotation=90)
plt.legend(title='After Waive Status')

# เพิ่มตัวเลขเปอร์เซ็นต์บนกราฟแท่ง
for container in ax.containers:
    for patch in container.patches:
        height = patch.get_height()
        if height > 0: # แสดงเฉพาะส่วนที่มีค่ามากกว่า 0
            ax.text(patch.get_x() + patch.get_width() / 2,
                    patch.get_y() + height / 2,
                    f'{height:.0f}%', # แสดงค่าเปอร์เซ็นต์แบบไม่มีทศนิยม
                    ha='center', va='center',
                    color='black', # ใช้สีดำเพื่อให้เห็นชัดเจนบนแท่งสีเข้ม
                    fontsize=10) # เพิ่มขนาดตัวอักษร

plt.tight_layout()
plt.show()
```

```python
df['FINISH_DATE'] = pd.to_datetime(df['FINISH_DATE'], errors='coerce')
df['ONSITE_DATE'] = pd.to_datetime(df['ONSITE_DATE'], errors='coerce')
df.info()
```

```python
df['Onsite_to_Done_Duration'] = df['FINISH_DATE'] - df['ONSITE_DATE']
df['Onsite_to_Done_Minutes'] = df['Onsite_to_Done_Duration'].dt.total_seconds() / 60
print("New DataFrame head with 'Onsite_to_Done_Duration' and 'Onsite_to_Done_Minutes':")
print(df[['ONSITE_DATE', 'FINISH_DATE', 'Onsite_to_Done_Duration', 'Onsite_to_Done_Minutes']].head())
```

```python
avg_onsite_to_done = df.groupby('ASSIGN_TO')['Onsite_to_Done_Minutes'].mean().sort_values(ascending=False)
print("Average 'Onsite to Done' duration (in minutes) per ASSIGN_TO:\n")
print(avg_onsite_to_done.head(10))
```

```python
plt.figure(figsize=(12, 10))
sns.barplot(x=avg_onsite_to_done.head(20).values, y=avg_onsite_to_done.head(20).index, palette='viridis', hue=avg_onsite_to_done.head(20).index, legend=False)
plt.title('Top 20 Average Onsite to Done Duration per ASSIGN_TO (Minutes)')
plt.xlabel('Average Duration (Minutes)')
plt.ylabel('ASSIGN_TO')
plt.tight_layout()
plt.show()
```

# เรียนรู้การใช้งาน Git & GitHub

1. รู้จักกับการจัดเก็บเวอร์ชั่น (Version Control)
- Version Control หรือ Source Control หมายถึงเครื่องมือช่วยติดตามการเปลี่ยนแปลงของ Source Code โดยการเก็บประวัติการเปลี่ยนแปลงลงฐานข้อมูลชนิดพิเศษ ซึ่งจุดประสงค์ของการเก็บบันทึกทุกการเปลี่ยนแปลงก็คือ หากมีความผลิดพลาดเกิดขึ้น ไม่ว่าจะผิดพลาดเล็กน้อยไปจนถึงผิดร้ายแรงจะส่งผลห้ Softwereที่พัฒนาขึ้นมานั้นพังทั้งหมด
- การบันทึกประวัติเก็บไว้อย่างต่อเนื่อง จะช่วยทำให้นักพัฒนาซอร์ฟแวร์สามารถย้อนกลับไปในช่วงเวลาต่างๆ โดยเปรียบเทียบโค้ดใหม่กับโค้ดอีกเวอร์ชั่นก่อนหน้าได้ และตรวจสอบความผิดพลาดเพื่อแก้ไขให้มีผลกระทบต่อทีมพัฒนาน้อยที่สุด คล้ายๆกับการ Undo/Redo ในโปรแกรม (แต่ถ้าหากปิดโปรแกรมไปก็จะไม่สามารถ Undo/Redo ได้อีกรอบ)

2. สาเหตุที่ต้องใช้ Version Control
  - Backup code ที่เขียนเพื่อให้สามารถกลับมาใช้งานได้ภายหลัง
  - คัดลอกไฟล์เก่าและสร้างไฟล์ใหม่ทุกครั้งที่มีการแก้ไข code
  - project แก้ปัญหาเรื่องของการสร้างโปรเจคหลาย folder เก็บ Code เวอร์ชั่นเก่าเอาไว้ กรณีเวอร์ชั่นใหม่พัง จะได้นำกลับมาใช้ใหม่ได้ ก็จะมีการ Backup เรื่อยๆ
  - แก้ไขปัญหาถ้าเป็นไฟล์โปรเจกต์ใหญ่  มีความซับซ้อนมีไฟล์เยอะๆ จะช่วยแก้ไขปัญหาตรงนี้ได้
  - แก้ปัญหากรณีกลับไปแก้งานเก่าที่ทำไวว้ จะต้องไปหาไฟล์ Backup แล้วจำได้ไหมว่าไฟล์ Backup นั้นถูกแก้อะไรบ้าง และเวลาเท่าไหร่
  - สามารถเพิ่มฟีเจอร์ใหม่ได้ โดยไม่ต้องเขียน code ใหม่
  - สามารถหาทีมช่วยทำได้ กรณีนำไฟล์ขึ้น github ช่วยแก้ไขปัญหาการส่งงาน คนในทีมแก้ไข แก้ไข code จุดใดบ้าง อยากพัฒนาระบบที่มีอยู่ไม่ให้กระทบกับ Production
  - สามารถติดตามการเปลี่ยนแปลง code ในไฟล์นั้นได้ รวมถึงย้อนเวลากลับไปก่อนที่ code จะพังได้

3. วิวัฒนาการของ Version Control
- **copy file & Folder**
ลักษณะการบันทึกเอกสารและแยกออกเป็นหลายไฟล์ แล้วตั้งชื่อไฟล์พร้อมระบุเวอร์ชั่นตามลำดับ




ข้อเสีย เปลืองพื้นที่การจัดเก็บข้อมูลเอกสาร เพราะต้องคัดลอกไฟล์ทั้งโฟลเดอร์เพื่อสร้างเอกสารเวอร์ชั่นใหม่
- **Patch**
เลี่ยงเรื่องเปลืองพื้นที่ในการจัดเก็บ คือการเปรียบเทียบไฟล์ใหม่กับเก่า นำมาเปรียบเทียบต่างกันยังไงแล้วจะได้ไฟล์ใหม่มา แต่ความเสี่ยงคือถ้า Patch หายจะไม่สามารถย้อนกลับไปได้ กระทบกับร่าง Source code
- **Local Version Control System**
ใช้ในการแก้ไขปัญหา Patch หาย จนไม่าสามารถประกรอบร่าง Source Code กลับไปยังเวอร์ชั่นต่างๆได้ จึงได้มีการพัฒน์ Version Control system(VCS) ที่มีฐานข้อมูลเฉพาะคอยจัดเก็บทุกการเปลี่ยนแปลงของ Source code โดยจะเรียกการจัดเกอร์เวอร์ชั่นของ Source code ลงฐานข้อมูลเรียกว่า "Check-in" และเรียกคืน Source code จากฐานข้อมูลเพื่อทำงานต่อว่า "Check-Out" สามารถ Log Message ว่าจะแก้ไขตรงส่วนใหน เพื่อให้นักพัฒนาซอฟแวร์แก้ไข Source code ต่อไปได้
ข้อเสีย การ Checkin out จากผู้ใช้หลายๆคนอาจจะทำให้เกิดปัญหาขึ้นได้
- **Centralized Version Control System(CVCS)**
เป็นเวอร์ชั่นคอนโทรล แบบรวมศูนย์ เพื่อแก้ปัญหากรณีที่มีผู้ใช้งานหลายคนโดยการเก็บข้อมูลไว้บน Server ซึ่งเมื่อผู้ใช้ Checkout งานเวอร์ชั่นเดียวกันแล้วกลับมา Check-in ระบบจะพยายามรวมเนื้อหาเข้าด้วยกัน (Merge)
ถ้ารูปแบบการรวมเนื้อหาเดียวกันมีความเรียบง่ายระบบจะรวมเนื้อหาให้อัตโนมัติ
ถ้ามีความซับซ้อนจะให้ผู้ใช้งานตัดสินใจแทน
ข้อเสีย กรณี Server ล่ม จะไม่สามารถทำงานต่อได้ Update งานไม่ได้ ถ้าล่มถาวร Source จะได้รับความเสียหายด้วย และตต้องทำ Online เท่านั้น
- **Distributed Version Control System (DVCS)**
จะแก้ปัญหาแบบรวมศูนย์โดยการ Clone ฐานข้อมูลมาทั้ง จะสามารถเช็คอิน เอาท์แบบ offline ได้เลย  สามารถดึงโปรเจคจาก Server เข้ามาเขียน Offline ได้
(เรียก database ที่เก็บ​Sourceว่า Repository)
ไม่ได้มีการรวมศูนย์ สามารถโคลนโปรเจคได้
ซึ่ง Git และ Git Hub คือตัวนี้


4. รูปจักกับ Git และ Github

- Git เป็น Version Control รูปแบบ Distributed Version Control System DVCS ใช้สำหรับการติดตาม ตรวจสอบการแก้ไขไฟล์ หรือ Source Code ซึ่งสามรรถตรวจสอบได้ทุกตัวอักษร ทุกบรรทัด ทุกไฟล์ พร้อมทั้งระบุว่าไฟล์ดังกล่าวถูกปรับปรุงช่วงเวลาใด ใครเป็นผู้แก้ไข
จะทำให้ผู้พัฒนาสามารถติดตามการเปลี่ยนแปลงได้ตลอด ย้อนกลับไปก่อนพังก็สามารถทำได้ ย้อนกลับไปได้
- ดังนั้น git จึงเหมาะกับนักพัฒนาที่ทำงานคนเดียวหรือเป็นทีม ถ้าทำเป็นทีมเราเรียกดูได้ว่าใครเป็นเขียนตัว code ตัวนี้เข้ามา เพราะต้องแบ่งงานทำ git สามารถตรวจสอบได้
- สามารถคัดลอก Version ออกมาในเครื่องได้ ทำให้การทำงานเร็วยิ่งขึ้น
ผู้พัฒนาสามารถแก้ไข code ได้ทุกที่แบบ offline ก็ได้ (local repository) เปรียบเสมือนฐานข้อมูลพิเศษ
ถ้าต้องการ Update ใหม่ก็เทียบ Local Repository กับ Remote Repository เมื่อแก้ไขหรือ update code จะทำการ sync ด้วยวิธี Pull Merge Pull.
- ติดตามเวอร์ชันก์ของ Code ได้
- เครื่องมืออำนวยความสะดวกสำหรับการทำงานเป็นทีม สามารถเก็บบันทึกการเปลี่ยนแปลงของ Source Cdoe เวอร์ชั่นล่าสุดไว้ที่ Local Repository ซึ่งไม่จำเป็นต้องต่ออินเตอร์เน็ตในการทำงาน

### สรุปคำสั่ง Git พื้นฐานที่ควรรู้
การใช้งาน Git มีคำสั่งพื้นฐานหลายคำสั่งที่สำคัญสำหรับการจัดการเวอร์ชันโค้ดของคุณ นี่คือคำสั่งหลักๆ ที่คุณจะใช้บ่อยที่สุด:

*   **`git init`**: ใช้สำหรับเริ่มต้น Git Repository ใหม่ในโปรเจกต์ของคุณ คำสั่งนี้จะสร้างโฟลเดอร์ `.git` ที่ซ่อนอยู่ ซึ่งเป็นที่เก็บข้อมูลประวัติเวอร์ชันทั้งหมดของโปรเจกต์

    ```bash
    git init
    ```

*   **`git add <file>` / `git add .`**: ใช้สำหรับเพิ่มไฟล์หรือโฟลเดอร์ไปยัง Staging Area (พื้นที่เตรียมพร้อม) ก่อนที่จะทำการบันทึกการเปลี่ยนแปลง (commit)
    *   `git add <file>`: เพิ่มไฟล์เดียว
    *   `git add .`: เพิ่มไฟล์ทั้งหมดที่มีการเปลี่ยนแปลงในไดเรกทอรีปัจจุบัน

    ```bash
    git add index.html
    git add .
    ```

*   **`git commit -m "<message>"`**: ใช้สำหรับบันทึกการเปลี่ยนแปลงที่อยู่ใน Staging Area เข้าสู่ Repository อย่างเป็นทางการ โดย `"<message>"` คือข้อความอธิบายการเปลี่ยนแปลงนั้นๆ ซึ่งควรจะสื่อความหมายและกระชับ

    ```bash
    git commit -m "Add initial HTML structure"
    ```

*   **`git status`**: ใช้สำหรับดูสถานะปัจจุบันของ Repository แสดงให้เห็นว่ามีไฟล์ใดบ้างที่ถูกเปลี่ยนแปลง, ไฟล์ใดอยู่ใน Staging Area, และไฟล์ใดที่ยังไม่ได้ถูกติดตาม

    ```bash
    git status
    ```

*   **`git log`**: ใช้สำหรับดูประวัติการ commit ทั้งหมดของโปรเจกต์ แสดงข้อมูลเช่น commit ID, ผู้เขียน, วันที่, และข้อความ commit

    ```bash
    git log
    ```

*   **`git diff`**: ใช้สำหรับดูความแตกต่างระหว่างไฟล์เวอร์ชันต่างๆ เช่น ระหว่างไฟล์ที่ยังไม่ได้ `add` กับไฟล์ที่อยู่ใน Staging Area หรือระหว่าง Staging Area กับ commit ล่าสุด

    ```bash
    git diff
    ```

*   **`git branch`**: ใช้สำหรับดูรายชื่อ branch ทั้งหมดที่มีใน Repository และ branch ที่คุณกำลังใช้งานอยู่

    ```bash
    git branch
    ```

*   **`git branch <branch_name>`**: ใช้สำหรับสร้าง branch ใหม่

    ```bash
    git branch feature/new-login
    ```

*   **`git checkout <branch_name>`**: ใช้สำหรับสลับไปทำงานบน branch อื่น หรือสลับไปยัง commit ที่ต้องการ

    ```bash
    git checkout main
    git checkout feature/new-login
    ```

*   **`git merge <branch_name>`**: ใช้สำหรับรวมการเปลี่ยนแปลงจาก branch หนึ่งเข้ากับ branch ปัจจุบันที่คุณกำลังทำงานอยู่

    ```bash
    # สมมติว่าคุณอยู่ที่ branch 'main'
    git merge feature/new-login
    ```

*   **`git remote add origin <URL>`**: ใช้สำหรับเชื่อมโยง Local Repository ของคุณเข้ากับ Remote Repository บนแพลตฟอร์มอย่าง GitHub โดย `origin` คือชื่อย่อที่ใช้เรียก Remote Repository และ `<URL>` คือที่อยู่ของ Remote Repository นั้น

    ```bash
    git remote add origin https://github.com/your-username/your-repo.git
    ```

*   **`git push -u origin <branch_name>`**: ใช้สำหรับอัปโหลดการเปลี่ยนแปลงจาก Local Repository ของคุณไปยัง Remote Repository บน `origin` และ `branch_name` ที่ระบุ `-u` (หรือ `--set-upstream`) จะตั้งค่าให้ branch ปัจจุบันเชื่อมโยงกับ remote branch นั้นๆ ในครั้งแรกที่ push

    ```bash
    git push -u origin main
    ```

*   **`git pull origin <branch_name>`**: ใช้สำหรับดาวน์โหลดและรวมการเปลี่ยนแปลงจาก Remote Repository เข้าสู่ Local Repository ของคุณ

    ```bash
    git pull origin main
    ```

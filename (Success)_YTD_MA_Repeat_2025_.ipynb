```python
engineer_ma_counts = df['Engineer MA'].value_counts()
print(engineer_ma_counts.head())
```

```python
print(engineer_ma_counts)
```

```python
jobs_by_engineer_and_team = df.groupby(['Partner (group)', 'Engineer MA']).size()
print(jobs_by_engineer_and_team.head())
```

```python
jobs_df = jobs_by_engineer_and_team.reset_index(name='job_count')

unique_teams = jobs_df['Partner (group)'].unique()

for team in unique_teams:
    team_df = jobs_df[jobs_df['Partner (group)'] == team]

    plt.figure(figsize=(15, 7))
    plt.bar(team_df['Engineer MA'], team_df['job_count'], color='skyblue')

    plt.title(f'จำนวนงานของวิศวกรในทีม {team}', fontproperties=thai_font, fontsize=16)
    plt.xlabel('ชื่อวิศวกร', fontproperties=thai_font, fontsize=12)
    plt.ylabel('จำนวนงาน', fontproperties=thai_font, fontsize=12)
    plt.xticks(rotation=45, ha='right', fontproperties=thai_font, fontsize=10)
    plt.yticks(fontproperties=thai_font, fontsize=10)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
# Set rcParams for global font fallback, although explicit fontproperties will be used for specific text elements
plt.rcParams['font.family'] = ['Laksaman', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# Ensure engineer_ma_counts is sorted (value_counts() usually sorts descending by default)
engineer_ma_counts_sorted = engineer_ma_counts.sort_values(ascending=False)

plt.figure(figsize=(15, 8))
plt.bar(engineer_ma_counts_sorted.index, engineer_ma_counts_sorted.values, color='skyblue')

plt.xlabel('ชื่อวิศวกร', fontsize=12, fontproperties=thai_font)
plt.ylabel('จำนวนงาน', fontsize=12, fontproperties=thai_font)
plt.title('จำนวนงานของ Engineer MA แต่ละคน', fontsize=16, fontproperties=thai_font)

plt.xticks(rotation=90, ha='right', fontsize=10, fontproperties=thai_font)
plt.yticks(fontsize=10, fontproperties=thai_font)

plt.tight_layout()
plt.show()
```

```python
# Calculate relationship_counts with SOLUTION DETAIL (Repeat) as index
relationship_data = df.groupby('SOLUTION DETAIL (Repeat)')['Control&Uncontrol1'].value_counts().unstack(fill_value=0)

# Calculate total for each solution detail and sort
relationship_data['Total'] = relationship_data['Control'] + relationship_data['UnControl']
relationship_data_sorted = relationship_data.sort_values(by='Total', ascending=True)

# Display the sorted relationship data
display(relationship_data_sorted.drop(columns='Total'))
```

```python
import matplotlib.pyplot as plt

# Prepare data for plotting (drop 'Total' column)
plot_data = relationship_data_sorted.drop(columns='Total')

# Adjust figure size dynamically for better readability with many categories
num_categories = len(plot_data)
fig_height = max(8, num_categories * 0.5) # Minimum 8 inches, 0.5 inches per category for better spacing

plt.figure(figsize=(15, fig_height))

# Create a horizontal stacked bar chart
ax = plot_data.plot(kind='barh', stacked=True, figsize=(15, fig_height), cmap='viridis', ax=plt.gca())

plt.title('Relationship between Control/Uncontrol and Solution Details (Sorted by Total Count)', fontsize=14)
plt.xlabel('Count', fontsize=12)
plt.ylabel('Solution Detail', fontsize=12)
plt.legend(title='Status', bbox_to_anchor=(1.05, 1), loc='upper left')

# Add total count labels on the bars with no offset
for i, (total_count, control_val, uncontrol_val) in enumerate(zip(relationship_data_sorted['Total'], relationship_data_sorted['Control'], relationship_data_sorted['UnControl'])):
    ax.text(total_count, i, f'{total_count}', va='center', ha='left', fontsize=10) # Removed +5 offset

plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout(rect=[0, 0, 0.85, 1]) # Adjust layout to make space for the legend and labels
plt.show()
```

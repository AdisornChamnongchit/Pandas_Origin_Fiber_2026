```python
assign_to_counts = df['ASSIGN_TO'].value_counts()

plt.figure(figsize=(12, 6))
ax = sns.barplot(x=assign_to_counts.index, y=assign_to_counts.values, hue=assign_to_counts.index, palette='viridis', legend=False)
plt.xlabel('Assign To')
plt.ylabel('Number of Tasks')
plt.title('Number of Tasks per Assignee (YTD 2026)')
plt.xticks(rotation=90)

# Display the numerical values on the bars
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge')

plt.tight_layout()
plt.show()
```

```python
# Calculate total tasks per ASSIGN_TO
total_tasks_per_assignee = df['ASSIGN_TO'].value_counts().rename('Total Tasks')

# Filter for 'After Waive' == 'Pass'
passed_waive_assignee = df[df['After Waive'] == 'Pass']

# Count 'Pass' tasks per ASSIGN_TO
passed_tasks_per_assignee = passed_waive_assignee['ASSIGN_TO'].value_counts().rename('Passed Waive Tasks')

# Combine the two Series into a DataFrame
assignee_waive_analysis = pd.DataFrame({
    'Total Tasks': total_tasks_per_assignee,
    'Passed Waive Tasks': passed_tasks_per_assignee
}).fillna(0)

# Calculate the percentage
assignee_waive_analysis['Percentage Passed'] = (assignee_waive_analysis['Passed Waive Tasks'] / assignee_waive_analysis['Total Tasks']) * 100

display(assignee_waive_analysis.sort_values(by='Percentage Passed', ascending=False).head(10)) # Display top 10 for brevity
```

```python
plt.figure(figsize=(12, len(assignee_waive_analysis) * 0.3))

assignee_waive_analysis_sorted = assignee_waive_analysis.sort_values(by='Percentage Passed', ascending=False)

ax = sns.barplot(x='Percentage Passed', y=assignee_waive_analysis_sorted.index, data=assignee_waive_analysis_sorted, hue=assignee_waive_analysis_sorted.index, palette='viridis', legend=False)
plt.xlabel('Percentage Passed (%)')
plt.ylabel('Assign To')
plt.title('Percentage of Tasks Passed After Waive by Assignee (YTD 2026)')


for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%', label_type='edge', padding=3)

plt.tight_layout()
plt.show()
```

```python
# Calculate total tasks per WFM_COMPANY
total_tasks_per_company = df['WFM_COMPANY'].value_counts().rename('Total Tasks')

# Filter for 'After Waive' == 'Pass'
passed_waive = df[df['After Waive'] == 'Pass']

# Count 'Pass' tasks per WFM_COMPANY
passed_tasks_per_company = passed_waive['WFM_COMPANY'].value_counts().rename('Passed Waive Tasks')

# Combine the two Series into a DataFrame
waive_analysis = pd.DataFrame({
    'Total Tasks': total_tasks_per_company,
    'Passed Waive Tasks': passed_tasks_per_company
}).fillna(0)

# Calculate the percentage
waive_analysis['Percentage Passed'] = (waive_analysis['Passed Waive Tasks'] / waive_analysis['Total Tasks']) * 100

display(waive_analysis.sort_values(by='Percentage Passed', ascending=False))
```

```python
plt.figure(figsize=(12, 7))
ax = sns.barplot(x=waive_analysis.index, y='Percentage Passed', data=waive_analysis.sort_values(by='Percentage Passed', ascending=False), hue=waive_analysis.index, palette='viridis', legend=False)
plt.xlabel('WFM Company')
plt.ylabel('Percentage Passed (%)')
plt.title('Percentage of Tasks Passed After Waive by WFM Company (YTD 2026)')
plt.xticks(rotation=45, ha='right')

# Display the percentage values on the bars
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%', label_type='edge')

plt.tight_layout()
plt.show()
```
